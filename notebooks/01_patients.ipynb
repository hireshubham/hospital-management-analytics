## 4. Data Cleaning

```python
df.head()
```

### 4.1 Gender

```python
Gender_cleaned = df["Gender"].str.strip().str.lower()
```

```python
Gender_cleaned.unique()
```

```python
Gender_cleaned = Gender_cleaned.replace(
    {
        "male": "Male",
        "female": "Female",
        "m": "Male",
        "f": "Female",
        "other": "Other"
    }
)
```

```python
display(Gender_cleaned.unique())
display(Gender_cleaned.nunique())
```

```python
Gender_cleaned.isna().sum()
```

```python
Gender_cleaned.value_counts(dropna = False)
```

```python
df["Gender"] = Gender_cleaned
```

```python
df["Gender"].unique()
```

### 4.2 Blood Group

```python
df["Blood_Group"].value_counts(dropna = False)
```

```python
df["Blood_Group"].unique()
```

```python
df["Blood_Group"].isna().sum()
```

### 4.3 City

```python
city_cleaned = df["City"].str.strip().str.lower()
city_cleaned.unique()
```

```python
city_cleaned = city_cleaned.str.title()
city_cleaned.unique()
```

```python
city_cleaned.isna().sum()
```

```python
df["City"] = city_cleaned
```

```python
df["City"].unique()
```

### 4.4 Insurance Type

```python
display(df["Insurance_Type"].unique())
display(df["Insurance_Type"].nunique())
```

```python
Insurance_cleaned = df["Insurance_Type"].str.strip().str.title()
Insurance_cleaned
```

```python
Insurance_cleaned.unique()
```

```python
Insurance_cleaned = Insurance_cleaned.replace(
    {
        "Govt": "Government"
    }
)
```

```python
display(Insurance_cleaned.unique())
display(Insurance_cleaned.nunique())
```

```python
Insurance_cleaned.isna().sum()
```

```python
df["Insurance_Type"] = Insurance_cleaned
df["Insurance_Type"].unique()
```

### 4.5 DOB

```python
df["DOB"] = dob_converted
display(df["DOB"].isna().sum())
display(df["DOB"].dtype)
```

### 4.6 Registration Date

```python
registration_converted.dtype
```

```python
registration_converted.isna().sum()
```

```python
df["Registration_Date"] = registration_converted
df["Registration_Date"].dtype
```
